````python
import os
import json
from pathlib import Path

# Створюємо базову структуру проекту
project_root = Path('/home/user/output/cheatsheet-hub')
project_root.mkdir(exist_ok=True)

# Основні директорії проекту
directories = [
    'src/app',
    'src/components/ui',
    'src/components/layout',
    'src/components/search',
    'src/components/cheatsheet',
    'src/lib',
    'src/types',
    'src/data',
    'src/styles',
    'public/icons',
    'public/images'
]

for directory in directories:
    (project_root / directory).mkdir(parents=True, exist_ok=True)

# package.json
package_json = {
    "name": "cheatsheet-hub",
    "version": "0.1.0",
    "private": True,
    "scripts": {
        "dev": "next dev",
        "build": "next build",
        "start": "next start",
        "lint": "next lint",
        "type-check": "tsc --noEmit"
    },
    "dependencies": {
        "next": "14.0.0",
        "react": "^18.2.0",
        "react-dom": "^18.2.0",
        "lucide-react": "^0.292.0",
        "clsx": "^2.0.0",
        "tailwind-merge": "^2.0.0",
        "prismjs": "^1.29.0",
        "react-hot-toast": "^2.4.1"
    },
    "devDependencies": {
        "typescript": "^5.2.2",
        "@types/node": "^20.8.0",
        "@types/react": "^18.2.0",
        "@types/react-dom": "^18.2.0",
        "@types/prismjs": "^1.26.3",
        "autoprefixer": "^10.4.16",
        "eslint": "^8.51.0",
        "eslint-config-next": "14.0.0",
        "postcss": "^8.4.31",
        "tailwindcss": "^3.3.5"
    }
}

with open(project_root / 'package.json', 'w', encoding='utf-8') as f:
    json.dump(package_json, f, indent=2, ensure_ascii=False)

# tsconfig.json
tsconfig = {
    "compilerOptions": {
        "target": "es5",
        "lib": ["dom", "dom.iterable", "es6"],
        "allowJs": True,
        "skipLibCheck": True,
        "strict": True,
        "noEmit": True,
        "esModuleInterop": True,
        "module": "esnext",
        "moduleResolution": "bundler",
        "resolveJsonModule": True,
        "isolatedModules": True,
        "jsx": "preserve",
        "incremental": True,
        "plugins": [
            {
                "name": "next"
            }
        ],
        "baseUrl": ".",
        "paths": {
            "@/*": ["./src/*"],
            "@/components/*": ["./src/components/*"],
            "@/lib/*": ["./src/lib/*"],
            "@/types/*": ["./src/types/*"],
            "@/data/*": ["./src/data/*"]
        }
    },
    "include": ["next-env.d.ts", "**/*.ts", "**/*.tsx", ".next/types/**/*.ts"],
    "exclude": ["node_modules"]
}

with open(project_root / 'tsconfig.json', 'w', encoding='utf-8') as f:
    json.dump(tsconfig, f, indent=2, ensure_ascii=False)

# tailwind.config.js
tailwind_config = '''/** @type {import('tailwindcss').Config} */
module.exports = {
  content: [
    './src/pages/**/*.{js,ts,jsx,tsx,mdx}',
    './src/components/**/*.{js,ts,jsx,tsx,mdx}',
    './src/app/**/*.{js,ts,jsx,tsx,mdx}',
  ],
  darkMode: 'class',
  theme: {
    extend: {
      colors: {
        border: 'hsl(var(--border))',
        input: 'hsl(var(--input))',
        ring: 'hsl(var(--ring))',
        background: 'hsl(var(--background))',
        foreground: 'hsl(var(--foreground))',
        primary: {
          DEFAULT: 'hsl(var(--primary))',
          foreground: 'hsl(var(--primary-foreground))',
        },
        secondary: {
          DEFAULT: 'hsl(var(--secondary))',
          foreground: 'hsl(var(--secondary-foreground))',
        },
        destructive: {
          DEFAULT: 'hsl(var(--destructive))',
          foreground: 'hsl(var(--destructive-foreground))',
        },
        muted: {
          DEFAULT: 'hsl(var(--muted))',
          foreground: 'hsl(var(--muted-foreground))',
        },
        accent: {
          DEFAULT: 'hsl(var(--accent))',
          foreground: 'hsl(var(--accent-foreground))',
        },
        popover: {
          DEFAULT: 'hsl(var(--popover))',
          foreground: 'hsl(var(--popover-foreground))',
        },
        card: {
          DEFAULT: 'hsl(var(--card))',
          foreground: 'hsl(var(--card-foreground))',
        },
      },
      borderRadius: {
        lg: 'var(--radius)',
        md: 'calc(var(--radius) - 2px)',
        sm: 'calc(var(--radius) - 4px)',
      },
      fontFamily: {
        mono: ['JetBrains Mono', 'Fira Code', 'Consolas', 'monospace'],
      },
    },
  },
  plugins: [],
}'''

with open(project_root / 'tailwind.config.js', 'w', encoding='utf-8') as f:
    f.write(tailwind_config)

# postcss.config.js
postcss_config = '''module.exports = {
  plugins: {
    tailwindcss: {},
    autoprefixer: {},
  },
}'''

with open(project_root / 'postcss.config.js', 'w', encoding='utf-8') as f:
    f.write(postcss_config)

# next.config.js
next_config = '''/** @type {import('next').NextConfig} */
const nextConfig = {
  experimental: {
    appDir: true,
  },
  images: {
    domains: ['localhost'],
  },
}

module.exports = nextConfig'''

with open(project_root / 'next.config.js', 'w', encoding='utf-8') as f:
    f.write(next_config)

# .eslintrc.json
eslint_config = {
    "extends": ["next/core-web-vitals", "next/typescript"],
    "rules": {
        "@typescript-eslint/no-unused-vars": "warn",
        "react-hooks/exhaustive-deps": "warn"
    }
}

with open(project_root / '.eslintrc.json', 'w', encoding='utf-8') as f:
    json.dump(eslint_config, f, indent=2)

# .gitignore
gitignore_content = '''# See https://help.github.com/articles/ignoring-files/ for more about ignoring files.

# dependencies
/node_modules
/.pnp
.pnp.js

# testing
/coverage

# next.js
/.next/
/out/

# production
/build

# misc
.DS_Store
*.pem

# debug
npm-debug.log*
yarn-debug.log*
yarn-error.log*

# local env files
.env*.local

# vercel
.vercel

# typescript
*.tsbuildinfo
next-env.d.ts'''

with open(project_root / '.gitignore', 'w', encoding='utf-8') as f:
    f.write(gitignore_content)

# README.md
readme_content = '''# CheatSheet Hub

Повноцінний хаб шпаргалок для системних адміністраторів і програмістів.

## Особливості

- 🔍 Потужний пошук по назві та тегах
- 📂 Організація по категоріях
- 🎨 Підсвічування синтаксису коду
- 📱 Адаптивний дизайн
- 🌙 Темна/світла тема
- 📋 Копіювання коду в буфер обміну

## Технології

- Next.js 14 з App Router
- TypeScript
- Tailwind CSS
- Prism.js для підсвічування коду
- Lucide React для іконок

## Встановлення

```bash
npm install
npm run dev
```

Відкрийте [http://localhost:3000](http://localhost:3000) у браузері.

## Структура проекту

```
src/
├── app/                 # App Router сторінки
├── components/          # React компоненти
│   ├── ui/             # Базові UI компоненти
│   ├── layout/         # Компоненти макету
│   ├── search/         # Компоненти пошуку
│   └── cheatsheet/     # Компоненти шпаргалок
├── lib/                # Утиліти та хелпери
├── types/              # TypeScript типи
├── data/               # Дані шпаргалок
└── styles/             # Глобальні стилі
```

## Розробка

- `npm run dev` - запуск в режимі розробки
- `npm run build` - збірка для продакшену
- `npm run start` - запуск продакшен збірки
- `npm run lint` - перевірка коду
- `npm run type-check` - перевірка типів TypeScript
'''

with open(project_root / 'README.md', 'w', encoding='utf-8') as f:
    f.write(readme_content)

# Створюємо базові типи TypeScript
types_content = '''export interface CheatSheet {
  id: string;
  title: string;
  description: string;
  category: string;
  tags: string[];
  content: CheatSheetSection[];
  createdAt: string;
  updatedAt: string;
  difficulty: 'beginner' | 'intermediate' | 'advanced';
  language?: string;
}

export interface CheatSheetSection {
  id: string;
  title: string;
  description?: string;
  code: string;
  language: string;
  explanation?: string;
}

export interface Category {
  id: string;
  name: string;
  description: string;
  icon: string;
  color: string;
  count: number;
}

export interface SearchFilters {
  category?: string;
  tags?: string[];
  difficulty?: string;
  language?: string;
}

export interface SearchResult {
  cheatsheets: CheatSheet[];
  totalCount: number;
  categories: Category[];
}

export type Theme = 'light' | 'dark';

export interface AppState {
  theme: Theme;
  searchQuery: string;
  filters: SearchFilters;
  favorites: string[];
}'''

with open(project_root / 'src/types/index.ts', 'w', encoding='utf-8') as f:
    f.write(types_content)

# Створюємо глобальні стилі
globals_css = '''@tailwind base;
@tailwind components;
@tailwind utilities;

@layer base {
  :root {
    --background: 0 0% 100%;
    --foreground: 222.2 84% 4.9%;
    --card: 0 0% 100%;
    --card-foreground: 222.2 84% 4.9%;
    --popover: 0 0% 100%;
    --popover-foreground: 222.2 84% 4.9%;
    --primary: 221.2 83.2% 53.3%;
    --primary-foreground: 210 40% 98%;
    --secondary: 210 40% 96%;
    --secondary-foreground: 222.2 84% 4.9%;
    --muted: 210 40% 96%;
    --muted-foreground: 215.4 16.3% 46.9%;
    --accent: 210 40% 96%;
    --accent-foreground: 222.2 84% 4.9%;
    --destructive: 0 84.2% 60.2%;
    --destructive-foreground: 210 40% 98%;
    --border: 214.3 31.8% 91.4%;
    --input: 214.3 31.8% 91.4%;
    --ring: 221.2 83.2% 53.3%;
    --radius: 0.5rem;
  }

  .dark {
    --background: 222.2 84% 4.9%;
    --foreground: 210 40% 98%;
    --card: 222.2 84% 4.9%;
    --card-foreground: 210 40% 98%;
    --popover: 222.2 84% 4.9%;
    --popover-foreground: 210 40% 98%;
    --primary: 217.2 91.2% 59.8%;
    --primary-foreground: 222.2 84% 4.9%;
    --secondary: 217.2 32.6% 17.5%;
    --secondary-foreground: 210 40% 98%;
    --muted: 217.2 32.6% 17.5%;
    --muted-foreground: 215 20.2% 65.1%;
    --accent: 217.2 32.6% 17.5%;
    --accent-foreground: 210 40% 98%;
    --destructive: 0 62.8% 30.6%;
    --destructive-foreground: 210 40% 98%;
    --border: 217.2 32.6% 17.5%;
    --input: 217.2 32.6% 17.5%;
    --ring: 224.3 76.3% 94.1%;
  }
}

@layer base {
  * {
    @apply border-border;
  }
  body {
    @apply bg-background text-foreground;
  }
}

/* Prism.js стилі для підсвічування коду */
pre[class*="language-"] {
  @apply rounded-lg p-4 overflow-x-auto;
}

code[class*="language-"] {
  @apply text-sm font-mono;
}

/* Кастомні стилі для компонентів */
.cheatsheet-card {
  @apply bg-card border border-border rounded-lg p-6 hover:shadow-lg transition-shadow;
}

.code-block {
  @apply bg-muted rounded-md p-4 font-mono text-sm overflow-x-auto;
}

.search-input {
  @apply w-full px-4 py-2 border border-input rounded-md bg-background text-foreground placeholder:text-muted-foreground focus:outline-none focus:ring-2 focus:ring-ring;
}

.category-badge {
  @apply inline-flex items-center px-2.5 py-0.5 rounded-full text-xs font-medium bg-primary/10 text-primary;
}

.tag-badge {
  @apply inline-flex items-center px-2 py-1 rounded-md text-xs font-medium bg-secondary text-secondary-foreground;
}'''

with open(project_root / 'src/styles/globals.css', 'w', encoding='utf-8') as f:
    f.write(globals_css)

# Створюємо утиліти
utils_content = '''import { type ClassValue, clsx } from "clsx"
import { twMerge } from "tailwind-merge"

export function cn(...inputs: ClassValue[]) {
  return twMerge(clsx(inputs))
}

export function formatDate(date: string): string {
  return new Date(date).toLocaleDateString('uk-UA', {
    year: 'numeric',
    month: 'long',
    day: 'numeric'
  })
}

export function copyToClipboard(text: string): Promise<void> {
  return navigator.clipboard.writeText(text)
}

export function debounce<T extends (...args: any[]) => any>(
  func: T,
  wait: number
): (...args: Parameters<T>) => void {
  let timeout: NodeJS.Timeout
  return (...args: Parameters<T>) => {
    clearTimeout(timeout)
    timeout = setTimeout(() => func(...args), wait)
  }
}

export function slugify(text: string): string {
  return text
    .toLowerCase()
    .replace(/[^\w\s-]/g, '')
    .replace(/[\s_-]+/g, '-')
    .replace(/^-+|-+$/g, '')
}

export function highlightSearchTerm(text: string, searchTerm: string): string {
  if (!searchTerm) return text
  
  const regex = new RegExp(`(${searchTerm})`, 'gi')
  return text.replace(regex, '<mark>$1</mark>')
}'''

with open(project_root / 'src/lib/utils.ts', 'w', encoding='utf-8') as f:
    f.write(utils_content)

print("✅ Базова структура Next.js проекту створена успішно!")
print(f"📁 Проект створено в: {project_root}")
print("\n📋 Створені файли та директорії:")
print("├── package.json (з усіма залежностями)")
print("├── tsconfig.json (конфігурація TypeScript)")
print("├── tailwind.config.js (конфігурація Tailwind CSS)")
print("├── next.config.js (конфігурація Next.js)")
print("├── .eslintrc.json (правила ESLint)")
print("├── .gitignore")
print("├── README.md")
print("└── src/")
print("    ├── types/index.ts (TypeScript типи)")
print("    ├── lib/utils.ts (утиліти)")
print("    └── styles/globals.css (глобальні стилі)")
print("\n🚀 Наступний крок: створення компонентів та сторінок")
````

````python
# Завершуємо створення JSON структури та утиліт
project_root = Path('/home/user/output/cheatsheet-hub')
data_dir = project_root / 'src/data'
lib_dir = project_root / 'src/lib'

# Переконуємося, що всі директорії існують
data_dir.mkdir(parents=True, exist_ok=True)
lib_dir.mkdir(parents=True, exist_ok=True)

# Завершуємо файл з прикладами використання даних
examples_content = '''// Приклади використання даних CheatSheet Hub

import { categories, cheatsheets, tags } from './index';
import type { CheatSheet, Category } from '../types';

// 1. Отримати всі категорії з підрахунком шпаргалок
export const getCategoriesWithCount = (): Category[] => {
  const categoryCount = new Map<string, number>();
  
  cheatsheets.forEach(sheet => {
    categoryCount.set(sheet.category, (categoryCount.get(sheet.category) || 0) + 1);
  });
  
  return categories.map(category => ({
    ...category,
    count: categoryCount.get(category.id) || 0
  }));
};

// 2. Знайти шпаргалки за категорією
export const getCheatSheetsByCategory = (categoryId: string): CheatSheet[] => {
  return cheatsheets.filter(sheet => sheet.category === categoryId);
};

// 3. Пошук шпаргалок за текстом
export const searchCheatSheets = (query: string): CheatSheet[] => {
  const searchTerm = query.toLowerCase();
  
  return cheatsheets.filter(sheet => 
    sheet.title.toLowerCase().includes(searchTerm) ||
    sheet.description.toLowerCase().includes(searchTerm) ||
    sheet.tags.some(tag => tag.toLowerCase().includes(searchTerm)) ||
    sheet.content.some(section => 
      section.title.toLowerCase().includes(searchTerm) ||
      section.code.toLowerCase().includes(searchTerm)
    )
  );
};

// 4. Отримати популярні теги
export const getPopularTags = (limit: number = 10): string[] => {
  const tagCount = new Map<string, number>();
  
  cheatsheets.forEach(sheet => {
    sheet.tags.forEach(tag => {
      tagCount.set(tag, (tagCount.get(tag) || 0) + 1);
    });
  });
  
  return Array.from(tagCount.entries())
    .sort((a, b) => b[1] - a[1])
    .slice(0, limit)
    .map(([tag]) => tag);
};

// 5. Отримати схожі шпаргалки
export const getSimilarCheatSheets = (cheatSheetId: string, limit: number = 5): CheatSheet[] => {
  const currentSheet = cheatsheets.find(sheet => sheet.id === cheatSheetId);
  if (!currentSheet) return [];
  
  const similar = cheatsheets
    .filter(sheet => sheet.id !== cheatSheetId)
    .map(sheet => ({
      sheet,
      similarity: calculateSimilarity(currentSheet, sheet)
    }))
    .sort((a, b) => b.similarity - a.similarity)
    .slice(0, limit)
    .map(item => item.sheet);
    
  return similar;
};

// Допоміжна функція для розрахунку схожості
const calculateSimilarity = (sheet1: CheatSheet, sheet2: CheatSheet): number => {
  let score = 0;
  
  // Схожість за категорією
  if (sheet1.category === sheet2.category) score += 3;
  
  // Схожість за тегами
  const commonTags = sheet1.tags.filter(tag => sheet2.tags.includes(tag));
  score += commonTags.length * 2;
  
  // Схожість за складністю
  if (sheet1.difficulty === sheet2.difficulty) score += 1;
  
  // Схожість за мовою програмування
  if (sheet1.language && sheet2.language && sheet1.language === sheet2.language) {
    score += 2;
  }
  
  return score;
};

// 6. Фільтрація за складністю
export const getCheatSheetsByDifficulty = (difficulty: 'beginner' | 'intermediate' | 'advanced'): CheatSheet[] => {
  return cheatsheets.filter(sheet => sheet.difficulty === difficulty);
};

// 7. Отримати останні оновлені шпаргалки
export const getRecentlyUpdated = (limit: number = 10): CheatSheet[] => {
  return [...cheatsheets]
    .sort((a, b) => new Date(b.updatedAt).getTime() - new Date(a.updatedAt).getTime())
    .slice(0, limit);
};

// 8. Статистика по мовах програмування
export const getLanguageStats = (): Record<string, number> => {
  const languageCount = new Map<string, number>();
  
  cheatsheets.forEach(sheet => {
    if (sheet.language) {
      languageCount.set(sheet.language, (languageCount.get(sheet.language) || 0) + 1);
    }
  });
  
  return Object.fromEntries(languageCount);
};

// 9. Експорт всіх утилітарних функцій
export const dataUtils = {
  getCategoriesWithCount,
  getCheatSheetsByCategory,
  searchCheatSheets,
  getPopularTags,
  getSimilarCheatSheets,
  getCheatSheetsByDifficulty,
  getRecentlyUpdated,
  getLanguageStats
};

// 10. Приклад використання всіх функцій
export const demonstrateUsage = () => {
  console.log('=== CheatSheet Hub Data Usage Examples ===');
  
  // Категорії з підрахунком
  const categoriesWithCount = getCategoriesWithCount();
  console.log('Категорії:', categoriesWithCount.map(c => `${c.name}: ${c.count}`));
  
  // Docker шпаргалки
  const dockerSheets = getCheatSheetsByCategory('docker');
  console.log(`Docker шпаргалок: ${dockerSheets.length}`);
  
  // Пошук
  const searchResults = searchCheatSheets('git');
  console.log(`Результатів пошуку "git": ${searchResults.length}`);
  
  // Популярні теги
  const popularTags = getPopularTags(5);
  console.log('Популярні теги:', popularTags);
  
  // Статистика мов
  const langStats = getLanguageStats();
  console.log('Статистика мов:', langStats);
  
  // Останні оновлення
  const recent = getRecentlyUpdated(3);
  console.log('Останні оновлення:', recent.map(s => s.title));
};'''

with open(data_dir / 'examples.ts', 'w', encoding='utf-8') as f:
    f.write(examples_content)

# Створюємо розширений сервіс для роботи з даними
advanced_service_content = '''import { CheatSheet, Category, SearchFilters, CheatSheetSection } from '@/types';
import { categories, cheatsheets, tags } from './index';

export interface SearchOptions {
  query?: string;
  category?: string;
  tags?: string[];
  difficulty?: 'beginner' | 'intermediate' | 'advanced';
  language?: string;
  sortBy?: 'title' | 'date' | 'difficulty' | 'category';
  sortOrder?: 'asc' | 'desc';
  limit?: number;
  offset?: number;
}

export interface SearchResult {
  items: CheatSheet[];
  total: number;
  hasMore: boolean;
  filters: {
    categories: Array<{ id: string; name: string; count: number }>;
    tags: Array<{ name: string; count: number }>;
    languages: Array<{ name: string; count: number }>;
    difficulties: Array<{ name: string; count: number }>;
  };
}

export class AdvancedCheatSheetService {
  private static instance: AdvancedCheatSheetService;
  private cheatsheets: CheatSheet[] = cheatsheets;
  private categories: Category[] = categories;

  static getInstance(): AdvancedCheatSheetService {
    if (!AdvancedCheatSheetService.instance) {
      AdvancedCheatSheetService.instance = new AdvancedCheatSheetService();
    }
    return AdvancedCheatSheetService.instance;
  }

  // Розширений пошук з фільтрами та пагінацією
  search(options: SearchOptions = {}): SearchResult {
    let results = [...this.cheatsheets];

    // Текстовий пошук
    if (options.query) {
      const query = options.query.toLowerCase();
      results = results.filter(sheet =>
        sheet.title.toLowerCase().includes(query) ||
        sheet.description.toLowerCase().includes(query) ||
        sheet.tags.some(tag => tag.toLowerCase().includes(query)) ||
        sheet.content.some(section =>
          section.title.toLowerCase().includes(query) ||
          section.code.toLowerCase().includes(query) ||
          (section.explanation && section.explanation.toLowerCase().includes(query))
        )
      );
    }

    // Фільтр за категорією
    if (options.category) {
      results = results.filter(sheet => sheet.category === options.category);
    }

    // Фільтр за тегами
    if (options.tags && options.tags.length > 0) {
      results = results.filter(sheet =>
        options.tags!.some(tag => sheet.tags.includes(tag))
      );
    }

    // Фільтр за складністю
    if (options.difficulty) {
      results = results.filter(sheet => sheet.difficulty === options.difficulty);
    }

    // Фільтр за мовою програмування
    if (options.language) {
      results = results.filter(sheet => 
        sheet.language?.toLowerCase() === options.language?.toLowerCase()
      );
    }

    // Сортування
    if (options.sortBy) {
      results.sort((a, b) => {
        let comparison = 0;
        
        switch (options.sortBy) {
          case 'title':
            comparison = a.title.localeCompare(b.title);
            break;
          case 'date':
            comparison = new Date(a.updatedAt).getTime() - new Date(b.updatedAt).getTime();
            break;
          case 'difficulty':
            const difficultyOrder = { beginner: 1, intermediate: 2, advanced: 3 };
            comparison = difficultyOrder[a.difficulty] - difficultyOrder[b.difficulty];
            break;
          case 'category':
            comparison = a.category.localeCompare(b.category);
            break;
        }
        
        return options.sortOrder === 'desc' ? -comparison : comparison;
      });
    }

    const total = results.length;

    // Пагінація
    if (options.offset || options.limit) {
      const offset = options.offset || 0;
      const limit = options.limit || 10;
      results = results.slice(offset, offset + limit);
    }

    // Генерація фільтрів для UI
    const filters = this.generateFilters(this.cheatsheets);

    return {
      items: results,
      total,
      hasMore: options.limit ? total > (options.offset || 0) + options.limit : false,
      filters
    };
  }

  // Генерація доступних фільтрів
  private generateFilters(sheets: CheatSheet[]) {
    const categoryCount = new Map<string, number>();
    const tagCount = new Map<string, number>();
    const languageCount = new Map<string, number>();
    const difficultyCount = new Map<string, number>();

    sheets.forEach(sheet => {
      // Підрахунок категорій
      categoryCount.set(sheet.category, (categoryCount.get(sheet.category) || 0) + 1);
      
      // Підрахунок тегів
      sheet.tags.forEach(tag => {
        tagCount.set(tag, (tagCount.get(tag) || 0) + 1);
      });
      
      // Підрахунок мов
      if (sheet.language) {
        languageCount.set(sheet.language, (languageCount.get(sheet.language) || 0) + 1);
      }
      
      // Підрахунок складності
      difficultyCount.set(sheet.difficulty, (difficultyCount.get(sheet.difficulty) || 0) + 1);
    });

    return {
      categories: this.categories.map(cat => ({
        id: cat.id,
        name: cat.name,
        count: categoryCount.get(cat.id) || 0
      })),
      tags: Array.from(tagCount.entries())
        .sort((a, b) => b[1] - a[1])
        .slice(0, 20)
        .map(([name, count]) => ({ name, count })),
      languages: Array.from(languageCount.entries())
        .sort((a, b) => b[1] - a[1])
        .map(([name, count]) => ({ name, count })),
      difficulties: Array.from(difficultyCount.entries())
        .map(([name, count]) => ({ name, count }))
    };
  }

  // Автодоповнення для пошуку
  getSearchSuggestions(query: string, limit: number = 10): string[] {
    const suggestions = new Set<string>();
    const searchTerm = query.toLowerCase();

    // Пошук в назвах
    this.cheatsheets.forEach(sheet => {
      if (sheet.title.toLowerCase().includes(searchTerm)) {
        suggestions.add(sheet.title);
      }
      
      // Пошук в тегах
      sheet.tags.forEach(tag => {
        if (tag.toLowerCase().includes(searchTerm)) {
          suggestions.add(tag);
        }
      });
    });

    return Array.from(suggestions).slice(0, limit);
  }

  // Експорт даних
  exportData(format: 'json' | 'csv' = 'json'): string {
    if (format === 'json') {
      return JSON.stringify({
        categories: this.categories,
        cheatsheets: this.cheatsheets,
        exportDate: new Date().toISOString(),
        version: '1.0.0'
      }, null, 2);
    }
    
    // CSV експорт (базовий)
    const headers = ['ID', 'Title', 'Category', 'Difficulty', 'Language', 'Tags', 'Created', 'Updated'];
    const rows = this.cheatsheets.map(sheet => [
      sheet.id,
      sheet.title,
      sheet.category,
      sheet.difficulty,
      sheet.language || '',
      sheet.tags.join(';'),
      sheet.createdAt,
      sheet.updatedAt
    ]);
    
    return [headers, ...rows].map(row => row.join(',')).join('\\n');
  }

  // Валідація даних
  validateData(): { isValid: boolean; errors: string[] } {
    const errors: string[] = [];

    // Перевірка унікальності ID
    const ids = new Set<string>();
    this.cheatsheets.forEach(sheet => {
      if (ids.has(sheet.id)) {
        errors.push(`Duplicate ID found: ${sheet.id}`);
      }
      ids.add(sheet.id);
    });

    // Перевірка існування категорій
    const categoryIds = new Set(this.categories.map(cat => cat.id));
    this.cheatsheets.forEach(sheet => {
      if (!categoryIds.has(sheet.category)) {
        errors.push(`Invalid category "${sheet.category}" in sheet "${sheet.id}"`);
      }
    });

    // Перевірка структури контенту
    this.cheatsheets.forEach(sheet => {
      if (!sheet.content || sheet.content.length === 0) {
        errors.push(`Empty content in sheet "${sheet.id}"`);
      }
      
      sheet.content.forEach((section, index) => {
        if (!section.code || section.code.trim() === '') {
          errors.push(`Empty code in section ${index} of sheet "${sheet.id}"`);
        }
      });
    });

    return {
      isValid: errors.length === 0,
      errors
    };
  }

  // Статистика
  getDetailedStatistics() {
    const stats = {
      totalSheets: this.cheatsheets.length,
      totalCategories: this.categories.length,
      totalSections: this.cheatsheets.reduce((sum, sheet) => sum + sheet.content.length, 0),
      averageSectionsPerSheet: 0,
      categoryDistribution: {} as Record<string, number>,
      difficultyDistribution: {} as Record<string, number>,
      languageDistribution: {} as Record<string, number>,
      tagDistribution: {} as Record<string, number>,
      creationTimeline: {} as Record<string, number>,
      updateTimeline: {} as Record<string, number>
    };

    stats.averageSectionsPerSheet = stats.totalSections / stats.totalSheets;

    // Розподіл по категоріях
    this.cheatsheets.forEach(sheet => {
      stats.categoryDistribution[sheet.category] = (stats.categoryDistribution[sheet.category] || 0) + 1;
      stats.difficultyDistribution[sheet.difficulty] = (stats.difficultyDistribution[sheet.difficulty] || 0) + 1;
      
      if (sheet.language) {
        stats.languageDistribution[sheet.language] = (stats.languageDistribution[sheet.language] || 0) + 1;
      }
      
      sheet.tags.forEach(tag => {
        stats.tagDistribution[tag] = (stats.tagDistribution[tag] || 0) + 1;
      });
      
      // Часові лінії
      const createdMonth = new Date(sheet.createdAt).toISOString().substring(0, 7);
      const updatedMonth = new Date(sheet.updatedAt).toISOString().substring(0, 7);
      
      stats.creationTimeline[createdMonth] = (stats.creationTimeline[createdMonth] || 0) + 1;
      stats.updateTimeline[updatedMonth] = (stats.updateTimeline[updatedMonth] || 0) + 1;
    });

    return stats;
  }
}

// Експорт singleton instance
export const advancedCheatSheetService = AdvancedCheatSheetService.getInstance();'''

with open(lib_dir / 'advanced-service.ts', 'w', encoding='utf-8') as f:
    f.write(advanced_service_content)

# Створюємо файл з тестовими даними та моками
test_data_content = '''import { CheatSheet, Category } from '@/types';

// Тестові дані для розробки та тестування
export const mockCategories: Category[] = [
  {
    id: 'test-category',
    name: 'Test Category',
    description: 'Category for testing purposes',
    icon: 'test',
    color: '#FF0000',
    count: 2
  }
];

export const mockCheatSheets: CheatSheet[] = [
  {
    id: 'test-sheet-1',
    title: 'Test CheatSheet 1',
    description: 'First test cheatsheet',
    category: 'test-category',
    tags: ['test', 'mock', 'example'],
    difficulty: 'beginner',
    language: 'javascript',
    createdAt: '2024-01-01T00:00:00.000Z',
    updatedAt: '2024-01-01T00:00:00.000Z',
    content: [
      {
        id: 'test-section-1',
        title: 'Test Section',
        description: 'A test section',
        code: 'console.log("Hello, World!");',
        language: 'javascript',
        explanation: 'This is a test code example'
      }
    ]
  },
  {
    id: 'test-sheet-2',
    title: 'Test CheatSheet 2',
    description: 'Second test cheatsheet',
    category: 'test-category',
    tags: ['test', 'advanced', 'example'],
    difficulty: 'advanced',
    language: 'python',
    createdAt: '2024-01-02T00:00:00.000Z',
    updatedAt: '2024-01-02T00:00:00.000Z',
    content: [
      {
        id: 'test-section-2',
        title: 'Advanced Test Section',
        description: 'An advanced test section',
        code: 'print("Hello, Python!")',
        language: 'python',
        explanation: 'This is an advanced test code example'
      }
    ]
  }
];

// Функції для генерації тестових даних
export const generateMockCheatSheet = (overrides: Partial<CheatSheet> = {}): CheatSheet => {
  const defaultSheet: CheatSheet = {
    id: `mock-${Date.now()}`,
    title: 'Mock CheatSheet',
    description: 'Generated mock cheatsheet for testing',
    category: 'test-category',
    tags: ['mock', 'generated'],
    difficulty: 'beginner',
    language: 'javascript',
    createdAt: new Date().toISOString(),
    updatedAt: new Date().toISOString(),
    content: [
      {
        id: `mock-section-${Date.now()}`,
        title: 'Mock Section',
        description: 'Generated mock section',
        code: '// Mock code example',
        language: 'javascript',
        explanation: 'This is a generated mock code example'
      }
    ]
  };

  return { ...defaultSheet, ...overrides };
};

export const generateMockCategory = (overrides: Partial<Category> = {}): Category => {
  const defaultCategory: Category = {
    id: `mock-category-${Date.now()}`,
    name: 'Mock Category',
    description: 'Generated mock category for testing',
    icon: 'test',
    color: '#000000',
    count: 0
  };

  return { ...defaultCategory, ...overrides };
};

// Утиліти для тестування
export const createTestEnvironment = () => {
  return {
    categories: [...mockCategories],
    cheatsheets: [...mockCheatSheets],
    addMockSheet: (sheet: Partial<CheatSheet>) => {
      const mockSheet = generateMockCheatSheet(sheet);
      mockCheatSheets.push(mockSheet);
      return mockSheet;
    },
    addMockCategory: (category: Partial<Category>) => {
      const mockCategory = generateMockCategory(category);
      mockCategories.push(mockCategory);
      return mockCategory;
    },
    reset: () => {
      mockCheatSheets.length = 0;
      mockCategories.length = 0;
      mockCheatSheets.push(...mockCheatSheets);
      mockCategories.push(...mockCategories);
    }
  };
};

// Експорт для тестів
export const testUtils = {
  mockCategories,
  mockCheatSheets,
  generateMockCheatSheet,
  generateMockCategory,
  createTestEnvironment
};'''

with open(data_dir / 'test-data.ts', 'w', encoding='utf-8') as f:
    f.write(test_data_content)

# Створюємо файл з міграціями даних
migrations_content = '''// Міграції для оновлення структури даних CheatSheet Hub

export interface Migration {
  version: string;
  description: string;
  up: (data: any) => any;
  down: (data: any) => any;
}

// Міграція 1.0.0 -> 1.1.0: Додавання поля difficulty
export const migration_1_1_0: Migration = {
  version: '1.1.0',
  description: 'Add difficulty field to cheatsheets',
  up: (data) => {
    if (data.cheatsheets) {
      data.cheatsheets = data.cheatsheets.map((sheet: any) => ({
        ...sheet,
        difficulty: sheet.difficulty || 'beginner'
      }));
    }
    return data;
  },
  down: (data) => {
    if (data.cheatsheets) {
      data.cheatsheets = data.cheatsheets.map((sheet: any) => {
        const { difficulty, ...rest } = sheet;
        return rest;
      });
    }
    return data;
  }
};

// Міграція 1.1.0 -> 1.2.0: Додавання поля language
export const migration_1_2_0: Migration = {
  version: '1.2.0',
  description: 'Add language field to cheatsheets',
  up: (data) => {
    if (data.cheatsheets) {
      data.cheatsheets = data.cheatsheets.map((sheet: any) => ({
        ...sheet,
        language: sheet.language || inferLanguageFromContent(sheet)
      }));
    }
    return data;
  },
  down: (data) => {
    if (data.cheatsheets) {
      data.cheatsheets = data.cheatsheets.map((sheet: any) => {
        const { language, ...rest } = sheet;
        return rest;
      });
    }
    return data;
  }
};

// Міграція 1.2.0 -> 1.3.0: Оновлення структури content
export const migration_1_3_0: Migration = {
  version: '1.3.0',
  description: 'Update content structure with explanations',
  up: (data) => {
    if (data.cheatsheets) {
      data.cheatsheets = data.cheatsheets.map((sheet: any) => ({
        ...sheet,
        content: sheet.content.map((section: any) => ({
          ...section,
          explanation: section.explanation || generateDefaultExplanation(section)
        }))
      }));
    }
    return data;
  },
  down: (data) => {
    if (data.cheatsheets) {
      data.cheatsheets = data.cheatsheets.map((sheet: any) => ({
        ...sheet,
        content: sheet.content.map((section: any) => {
          const { explanation, ...rest } = section;
          return rest;
        })
      }));
    }
    return data;
  }
};

// Допоміжні функції для міграцій
function inferLanguageFromContent(sheet: any): string {
  if (sheet.category === 'docker') return 'bash';
  if (sheet.category === 'git') return 'bash';
  if (sheet.category === 'linux') return 'bash';
  if (sheet.category === 'cloud') return 'bash';
  if (sheet.category === 'programming') {
    // Спробувати визначити мову з коду
    const codeContent = sheet.content?.[0]?.code || '';
    if (codeContent.includes('import ') && codeContent.includes('from ')) return 'python';
    if (codeContent.includes('const ') || codeContent.includes('let ')) return 'javascript';
    if (codeContent.includes('version:') && codeContent.includes('services:')) return 'yaml';
    return 'bash';
  }
  return 'bash';
}

function generateDefaultExplanation(section: any): string {
  return `Пояснення для секції "${section.title}": ${section.description || 'Основні команди та приклади використання.'}`;
}

// Менеджер міграцій
export class MigrationManager {
  private migrations: Migration[] = [
    migration_1_1_0,
    migration_1_2_0,
    migration_1_3_0
  ];

  // Застосувати всі міграції
  applyMigrations(data: any, fromVersion: string = '1.0.0', toVersion?: string): any {
    let currentData = { ...data };
    const targetVersion = toVersion || this.getLatestVersion();
    
    const applicableMigrations = this.migrations.filter(migration => 
      this.compareVersions(migration.version, fromVersion) > 0 &&
      this.compareVersions(migration.version, targetVersion) <= 0
    );

    applicableMigrations.forEach(migration => {
      console.log(`Applying migration ${migration.version}: ${migration.description}`);
      currentData = migration.up(currentData);
    });

    return {
      ...currentData,
      version: targetVersion,
      migratedAt: new Date().toISOString()
    };
  }

  // Відкатити міграції
  rollbackMigrations(data: any, fromVersion: string, toVersion: string): any {
    let currentData = { ...data };
    
    const rollbackMigrations = this.migrations
      .filter(migration => 
        this.compareVersions(migration.version, toVersion) > 0 &&
        this.compareVersions(migration.version, fromVersion) <= 0
      )
      .reverse();

    rollbackMigrations.forEach(migration => {
      console.log(`Rolling back migration ${migration.version}`);
      currentData = migration.down(currentData);
    });

    return {
      ...currentData,
      version: toVersion,
      rolledBackAt: new Date().toISOString()
    };
  }

  // Отримати останню версію
  getLatestVersion(): string {
    return this.migrations
      .map(m => m.version)
      .sort(this.compareVersions)
      .pop() || '1.0.0';
  }

  // Порівняти версії
  private compareVersions(a: string, b: string): number {
    const aParts = a.split('.').map(Number);
    const bParts = b.split('.').map(Number);
    
    for (let i = 0; i < Math.max(aParts.length, bParts.length); i++) {
      const aPart = aParts[i] || 0;
      const bPart = bParts[i] || 0;
      
      if (aPart > bPart) return 1;
      if (aPart < bPart) return -1;
    }
    
    return 0;
  }

  // Перевірити чи потрібні міграції
  needsMigration(currentVersion: string): boolean {
    const latestVersion = this.getLatestVersion();
    return this.compareVersions(latestVersion, current
````

````python
import os
from pathlib import Path

# Визначаємо шлях до проекту
project_root = Path('/home/user/output/cheatsheet-hub')
components_dir = project_root / 'src/components'

# Переконуємося, що всі директорії існують
for directory in ['search', 'cheatsheet', 'ui', 'layout']:
    (components_dir / directory).mkdir(parents=True, exist_ok=True)

# Завершуємо SearchResults компонент (виправляємо синтаксичну помилку)
search_results_component = '''import React from 'react';
import Link from 'next/link';
import { Card, CardContent, CardDescription, CardHeader, CardTitle } from '@/components/ui/card';
import { Badge } from '@/components/ui/badge';
import { Button } from '@/components/ui/button';
import { Heart, ExternalLink, Clock } from 'lucide-react';
import { useCheatSheetSearch, useFavorites } from '@/lib/hooks';
import { formatDate, getDifficultyColor } from '@/lib/data-utils';
import { CheatSheet } from '@/types';

interface SearchResultsProps {
  className?: string;
}

const SearchResults: React.FC<SearchResultsProps> = ({ className }) => {
  const { results, isLoading, totalResults } = useCheatSheetSearch();
  const { toggleFavorite, isFavorite } = useFavorites();

  if (isLoading) {
    return (
      <div className={`space-y-4 ${className}`}>
        {[...Array(3)].map((_, i) => (
          <Card key={i} className="animate-pulse">
            <CardHeader>
              <div className="h-4 bg-muted rounded w-3/4"></div>
              <div className="h-3 bg-muted rounded w-1/2"></div>
            </CardHeader>
            <CardContent>
              <div className="space-y-2">
                <div className="h-3 bg-muted rounded"></div>
                <div className="h-3 bg-muted rounded w-5/6"></div>
              </div>
            </CardContent>
          </Card>
        ))}
      </div>
    );
  }

  if (results.length === 0) {
    return (
      <div className={`text-center py-12 ${className}`}>
        <div className="text-muted-foreground">
          <h3 className="text-lg font-semibold mb-2">Нічого не знайдено</h3>
          <p>Спробуйте змінити пошуковий запит або фільтри</p>
        </div>
      </div>
    );
  }

  return (
    <div className={`space-y-4 ${className}`}>
      <div className="flex items-center justify-between">
        <p className="text-sm text-muted-foreground">
          Знайдено {totalResults} результатів
        </p>
      </div>

      <div className="grid gap-4">
        {results.map((cheatsheet: CheatSheet) => (
          <Card key={cheatsheet.id} className="hover:shadow-md transition-shadow">
            <CardHeader>
              <div className="flex items-start justify-between">
                <div className="space-y-1 flex-1">
                  <CardTitle className="text-lg">
                    <Link 
                      href={`/cheatsheet/${cheatsheet.id}`}
                      className="hover:text-primary transition-colors"
                    >
                      {cheatsheet.title}
                    </Link>
                  </CardTitle>
                  <CardDescription>{cheatsheet.description}</CardDescription>
                </div>
                <Button
                  variant="ghost"
                  size="icon"
                  onClick={() => toggleFavorite(cheatsheet.id)}
                  className={`ml-2 ${isFavorite(cheatsheet.id) ? 'text-red-500' : ''}`}
                >
                  <Heart className={`h-4 w-4 ${isFavorite(cheatsheet.id) ? 'fill-current' : ''}`} />
                </Button>
              </div>
            </CardHeader>
            <CardContent>
              <div className="space-y-3">
                {/* Tags and difficulty */}
                <div className="flex flex-wrap items-center gap-2">
                  <Badge 
                    variant="secondary" 
                    className={getDifficultyColor(cheatsheet.difficulty)}
                  >
                    {cheatsheet.difficulty}
                  </Badge>
                  {cheatsheet.tags.slice(0, 3).map(tag => (
                    <Badge key={tag} variant="outline" className="text-xs">
                      {tag}
                    </Badge>
                  ))}
                  {cheatsheet.tags.length > 3 && (
                    <Badge variant="outline" className="text-xs">
                      +{cheatsheet.tags.length - 3}
                    </Badge>
                  )}
                </div>

                {/* Metadata */}
                <div className="flex items-center justify-between text-xs text-muted-foreground">
                  <span>Категорія: {cheatsheet.category}</span>
                  <div className="flex items-center gap-1">
                    <Clock className="h-3 w-3" />
                    <span>Оновлено: {formatDate(cheatsheet.updatedAt)}</span>
                  </div>
                </div>

                {/* Preview of first section */}
                {cheatsheet.content && cheatsheet.content.length > 0 && (
                  <div className="mt-2 p-2 bg-muted rounded text-xs">
                    <div className="font-medium mb-1">{cheatsheet.content[0].title}</div>
                    <code className="text-muted-foreground">
                      {cheatsheet.content[0].code.split('\\n')[0]}...
                    </code>
                  </div>
                )}
              </div>
            </CardContent>
          </Card>
        ))}
      </div>
    </div>
  );
};

export default SearchResults;'''

with open(project_root / 'src/components/search/search-results.tsx', 'w', encoding='utf-8') as f:
    f.write(search_results_component)

# Search компонентів індекс
search_index = '''export { default as SearchBar } from './search-bar';
export { default as SearchFilters } from './search-filters';
export { default as SearchResults } from './search-results';'''

with open(project_root / 'src/components/search/index.ts', 'w', encoding='utf-8') as f:
    f.write(search_index)

# 4. Створюємо компоненти шпаргалок

# CheatSheetCard компонент
cheatsheet_card_component = '''import React from 'react';
import Link from 'next/link';
import { Card, CardContent, CardDescription, CardHeader, CardTitle } from '@/components/ui/card';
import { Badge } from '@/components/ui/badge';
import { Button } from '@/components/ui/button';
import { Heart, ExternalLink, Code, Calendar } from 'lucide-react';
import { CheatSheet } from '@/types';
import { useFavorites } from '@/lib/hooks';
import { formatDate, getDifficultyColor, getCategoryIcon } from '@/lib/data-utils';

interface CheatSheetCardProps {
  cheatsheet: CheatSheet;
  className?: string;
}

const CheatSheetCard: React.FC<CheatSheetCardProps> = ({ cheatsheet, className }) => {
  const { toggleFavorite, isFavorite } = useFavorites();

  const difficultyColors = {
    beginner: 'bg-green-100 text-green-800 dark:bg-green-900 dark:text-green-300',
    intermediate: 'bg-yellow-100 text-yellow-800 dark:bg-yellow-900 dark:text-yellow-300',
    advanced: 'bg-red-100 text-red-800 dark:bg-red-900 dark:text-red-300'
  };

  return (
    <Card className={`group hover:shadow-lg transition-all duration-200 ${className}`}>
      <CardHeader className="pb-3">
        <div className="flex items-start justify-between">
          <div className="space-y-1 flex-1">
            <CardTitle className="text-lg group-hover:text-primary transition-colors">
              <Link href={`/cheatsheet/${cheatsheet.id}`}>
                {cheatsheet.title}
              </Link>
            </CardTitle>
            <CardDescription className="line-clamp-2">
              {cheatsheet.description}
            </CardDescription>
          </div>
          <Button
            variant="ghost"
            size="icon"
            onClick={() => toggleFavorite(cheatsheet.id)}
            className={`ml-2 opacity-0 group-hover:opacity-100 transition-opacity ${
              isFavorite(cheatsheet.id) ? 'opacity-100 text-red-500' : ''
            }`}
          >
            <Heart className={`h-4 w-4 ${isFavorite(cheatsheet.id) ? 'fill-current' : ''}`} />
          </Button>
        </div>
      </CardHeader>

      <CardContent className="space-y-4">
        {/* Difficulty and Language */}
        <div className="flex items-center gap-2">
          <Badge className={difficultyColors[cheatsheet.difficulty]}>
            {cheatsheet.difficulty}
          </Badge>
          {cheatsheet.language && (
            <Badge variant="outline" className="gap-1">
              <Code className="h-3 w-3" />
              {cheatsheet.language}
            </Badge>
          )}
        </div>

        {/* Tags */}
        <div className="flex flex-wrap gap-1">
          {cheatsheet.tags.slice(0, 4).map(tag => (
            <Badge key={tag} variant="secondary" className="text-xs">
              {tag}
            </Badge>
          ))}
          {cheatsheet.tags.length > 4 && (
            <Badge variant="secondary" className="text-xs">
              +{cheatsheet.tags.length - 4}
            </Badge>
          )}
        </div>

        {/* Stats */}
        <div className="flex items-center justify-between text-xs text-muted-foreground">
          <div className="flex items-center gap-1">
            <Calendar className="h-3 w-3" />
            <span>{formatDate(cheatsheet.updatedAt)}</span>
          </div>
          <span>{cheatsheet.content.length} секцій</span>
        </div>

        {/* Preview */}
        {cheatsheet.content && cheatsheet.content.length > 0 && (
          <div className="bg-muted rounded-md p-3">
            <div className="text-xs font-medium text-muted-foreground mb-1">
              Попередній перегляд:
            </div>
            <code className="text-xs">
              {cheatsheet.content[0].code.split('\\n')[0].substring(0, 60)}
              {cheatsheet.content[0].code.length > 60 ? '...' : ''}
            </code>
          </div>
        )}

        {/* Action */}
        <Link href={`/cheatsheet/${cheatsheet.id}`}>
          <Button variant="outline" className="w-full group-hover:bg-primary group-hover:text-primary-foreground transition-colors">
            <ExternalLink className="h-4 w-4 mr-2" />
            Переглянути шпаргалку
          </Button>
        </Link>
      </CardContent>
    </Card>
  );
};

export default CheatSheetCard;'''

with open(project_root / 'src/components/cheatsheet/cheatsheet-card.tsx', 'w', encoding='utf-8') as f:
    f.write(cheatsheet_card_component)

# CodeBlock компонент
code_block_component = '''import React, { useState } from 'react';
import { Button } from '@/components/ui/button';
import { Badge } from '@/components/ui/badge';
import { Copy, Check } from 'lucide-react';
import { copyToClipboard } from '@/lib/utils';

interface CodeBlockProps {
  code: string;
  language: string;
  title?: string;
  explanation?: string;
  className?: string;
}

const CodeBlock: React.FC<CodeBlockProps> = ({
  code,
  language,
  title,
  explanation,
  className
}) => {
  const [copied, setCopied] = useState(false);

  const handleCopy = async () => {
    try {
      await copyToClipboard(code);
      setCopied(true);
      setTimeout(() => setCopied(false), 2000);
    } catch (error) {
      console.error('Failed to copy code:', error);
    }
  };

  return (
    <div className={`space-y-2 ${className}`}>
      {title && (
        <div className="flex items-center justify-between">
          <h4 className="text-sm font-semibold">{title}</h4>
          <Badge variant="outline" className="text-xs">
            {language}
          </Badge>
        </div>
      )}

      <div className="relative group">
        <pre className="bg-muted rounded-lg p-4 overflow-x-auto text-sm">
          <code className={`language-${language}`}>
            {code}
          </code>
        </pre>
        
        <Button
          variant="ghost"
          size="icon"
          onClick={handleCopy}
          className="absolute top-2 right-2 opacity-0 group-hover:opacity-100 transition-opacity"
        >
          {copied ? (
            <Check className="h-4 w-4 text-green-500" />
          ) : (
            <Copy className="h-4 w-4" />
          )}
        </Button>
      </div>

      {explanation && (
        <div className="text-sm text-muted-foreground bg-blue-50 dark:bg-blue-950 p-3 rounded-md">
          <strong>Пояснення:</strong> {explanation}
        </div>
      )}
    </div>
  );
};

export default CodeBlock;'''

with open(project_root / 'src/components/cheatsheet/code-block.tsx', 'w', encoding='utf-8') as f:
    f.write(code_block_component)

# CheatSheetContent компонент
cheatsheet_content_component = '''import React from 'react';
import { Card, CardContent, CardDescription, CardHeader, CardTitle } from '@/components/ui/card';
import { Badge } from '@/components/ui/badge';
import { Button } from '@/components/ui/button';
import { Heart, Share2, Download, Calendar, Tag, Code } from 'lucide-react';
import { CheatSheet } from '@/types';
import { useFavorites } from '@/lib/hooks';
import { formatDate } from '@/lib/data-utils';
import CodeBlock from './code-block';

interface CheatSheetContentProps {
  cheatsheet: CheatSheet;
  className?: string;
}

const CheatSheetContent: React.FC<CheatSheetContentProps> = ({ cheatsheet, className }) => {
  const { toggleFavorite, isFavorite } = useFavorites();

  const difficultyColors = {
    beginner: 'bg-green-100 text-green-800 dark:bg-green-900 dark:text-green-300',
    intermediate: 'bg-yellow-100 text-yellow-800 dark:bg-yellow-900 dark:text-yellow-300',
    advanced: 'bg-red-100 text-red-800 dark:bg-red-900 dark:text-red-300'
  };

  const handleShare = async () => {
    if (navigator.share) {
      try {
        await navigator.share({
          title: cheatsheet.title,
          text: cheatsheet.description,
          url: window.location.href
        });
      } catch (error) {
        console.log('Error sharing:', error);
      }
    } else {
      // Fallback to copying URL
      await navigator.clipboard.writeText(window.location.href);
    }
  };

  const handleDownload = () => {
    const content = cheatsheet.content
      .map(section => `# ${section.title}\\n\\n${section.code}\\n\\n`)
      .join('');
    
    const blob = new Blob([content], { type: 'text/plain' });
    const url = URL.createObjectURL(blob);
    const a = document.createElement('a');
    a.href = url;
    a.download = `${cheatsheet.title.replace(/\\s+/g, '_')}.txt`;
    document.body.appendChild(a);
    a.click();
    document.body.removeChild(a);
    URL.revokeObjectURL(url);
  };

  return (
    <div className={`space-y-6 ${className}`}>
      {/* Header */}
      <Card>
        <CardHeader>
          <div className="flex items-start justify-between">
            <div className="space-y-2 flex-1">
              <CardTitle className="text-2xl">{cheatsheet.title}</CardTitle>
              <CardDescription className="text-base">
                {cheatsheet.description}
              </CardDescription>
            </div>
            <div className="flex items-center gap-2 ml-4">
              <Button
                variant="outline"
                size="icon"
                onClick={() => toggleFavorite(cheatsheet.id)}
                className={isFavorite(cheatsheet.id) ? 'text-red-500' : ''}
              >
                <Heart className={`h-4 w-4 ${isFavorite(cheatsheet.id) ? 'fill-current' : ''}`} />
              </Button>
              <Button variant="outline" size="icon" onClick={handleShare}>
                <Share2 className="h-4 w-4" />
              </Button>
              <Button variant="outline" size="icon" onClick={handleDownload}>
                <Download className="h-4 w-4" />
              </Button>
            </div>
          </div>

          {/* Metadata */}
          <div className="flex flex-wrap items-center gap-3 pt-4">
            <Badge className={difficultyColors[cheatsheet.difficulty]}>
              {cheatsheet.difficulty}
            </Badge>
            
            {cheatsheet.language && (
              <Badge variant="outline" className="gap-1">
                <Code className="h-3 w-3" />
                {cheatsheet.language}
              </Badge>
            )}

            <div className="flex items-center gap-1 text-sm text-muted-foreground">
              <Calendar className="h-3 w-3" />
              <span>Оновлено: {formatDate(cheatsheet.updatedAt)}</span>
            </div>

            <div className="flex items-center gap-1 text-sm text-muted-foreground">
              <Tag className="h-3 w-3" />
              <span>{cheatsheet.content.length} секцій</span>
            </div>
          </div>

          {/* Tags */}
          <div className="flex flex-wrap gap-2 pt-2">
            {cheatsheet.tags.map(tag => (
              <Badge key={tag} variant="secondary" className="text-xs">
                {tag}
              </Badge>
            ))}
          </div>
        </CardHeader>
      </Card>

      {/* Content Sections */}
      <div className="space-y-6">
        {cheatsheet.content.map((section, index) => (
          <Card key={section.id}>
            <CardHeader>
              <CardTitle className="text-lg">{section.title}</CardTitle>
              {section.description && (
                <CardDescription>{section.description}</CardDescription>
              )}
            </CardHeader>
            <CardContent>
              <CodeBlock
                code={section.code}
                language={section.language}
                explanation={section.explanation}
              />
            </CardContent>
          </Card>
        ))}
      </div>
    </div>
  );
};

export default CheatSheetContent;'''

with open(project_root / 'src/components/cheatsheet/cheatsheet-content.tsx', 'w', encoding='utf-8') as f:
    f.write(cheatsheet_content_component)

# CheatSheetGrid компонент
cheatsheet_grid_component = '''import React from 'react';
import { CheatSheet } from '@/types';
import CheatSheetCard from './cheatsheet-card';

interface CheatSheetGridProps {
  cheatsheets: CheatSheet[];
  className?: string;
}

const CheatSheetGrid: React.FC<CheatSheetGridProps> = ({ cheatsheets, className }) => {
  if (cheatsheets.length === 0) {
    return (
      <div className={`text-center py-12 ${className}`}>
        <div className="text-muted-foreground">
          <h3 className="text-lg font-semibold mb-2">Шпаргалки не знайдені</h3>
          <p>Спробуйте змінити фільтри або пошуковий запит</p>
        </div>
      </div>
    );
  }

  return (
    <div className={`grid gap-6 md:grid-cols-2 lg:grid-cols-3 ${className}`}>
      {cheatsheets.map(cheatsheet => (
        <CheatSheetCard key={cheatsheet.id} cheatsheet={cheatsheet} />
      ))}
    </div>
  );
};

export default CheatSheetGrid;'''

with open(project_root / 'src/components/cheatsheet/cheatsheet-grid.tsx', 'w', encoding='utf-8') as f:
    f.write(cheatsheet_grid_component)

# SimilarCheatSheets компонент
similar_cheatsheets_component = '''import React from 'react';
import Link from 'next/link';
import { Card, CardContent, CardDescription, CardHeader, CardTitle } from '@/components/ui/card';
import { Badge } from '@/components/ui/badge';
import { CheatSheet } from '@/types';
import { ArrowRight } from 'lucide-react';

interface SimilarCheatSheetsProps {
  cheatsheets: CheatSheet[];
  className?: string;
}

const SimilarCheatSheets: React.FC<SimilarCheatSheetsProps> = ({ cheatsheets, className }) => {
  if (cheatsheets.length === 0) {
    return null;
  }

  return (
    <Card className={className}>
      <CardHeader>
        <CardTitle className="text-lg">Схожі шпаргалки</CardTitle>
        <CardDescription>
          Інші шпаргалки, які можуть вас зацікавити
        </CardDescription>
      </CardHeader>
      <CardContent>
        <div className="space-y-3">
          {cheatsheets.map(cheatsheet => (
            <Link
              key={cheatsheet.id}
              href={`/cheatsheet/${cheatsheet.id}`}
              className="block group"
            >
              <div className="flex items-center justify-between p-3 rounded-lg hover:bg-muted transition-colors">
                <div className="space-y-1 flex-1">
                  <div className="font-medium group-hover:text-primary transition-colors">
                    {cheatsheet.title}
                  </div>
                  <div className="text-sm text-muted-foreground line-clamp-1">
                    {cheatsheet.description}
                  </div>
                  <div className="flex gap-1">
                    {cheatsheet.tags.slice(0, 2).map(tag => (
                      <Badge key={tag} variant="secondary" className="text-xs">
                        {tag}
                      </Badge>
                    ))}
                  </div>
                </div>
                <ArrowRight className="h-4 w-4 text-muted-foreground group-hover:text-primary transition-colors" />
              </div>
            </Link>
          ))}
        </div>
      </CardContent>
    </Card>
  );
};

export default SimilarCheatSheets;'''

with open(project_root / 'src/components/cheatsheet/similar-cheatsheets.tsx', 'w', encoding='utf-8') as f:
    f.write(similar_cheatsheets_component)

# CategoryCard компонент
category_card_component = '''import React from 'react';
import Link from 'next/link';
import { Card, CardContent, CardDescription, CardHeader, CardTitle } from '@/components/ui/card';
import { Badge } from '@/components/ui/badge';
import { Category } from '@/types';
import { 
  Container, 
  GitBranch, 
  Terminal, 
  Cloud, 
  Code,
  FileText
} from 'lucide-react';

interface CategoryCardProps {
  category: Category;
  className?: string;
}

const CategoryCard: React.FC<CategoryCardProps> = ({ category, className }) => {
  const getIcon = (iconName: string) => {
    const icons = {
      container: Container,
      'git-branch': GitBranch,
      terminal: Terminal,
      cloud: Cloud,
      code: Code,
      default: FileText
    };
    
    const IconComponent = icons[iconName as keyof typeof icons] || icons.default;
    return <IconComponent className="h-8 w-8" />;
  };

  return (
    <Link href={`/category/${category.id}`}>
      <Card className={`group hover:shadow-lg transition-all duration-200 cursor-pointer ${className}`}>
        <CardHeader className="text-center">
          <div 
            className="mx-auto p-3 rounded-full w-fit group-hover:scale-110 transition-transform"
            style={{ backgroundColor: `${category.color}20`, color: category.color }}
          >
            {getIcon(category.icon)}
          </div>
          <CardTitle className="group-hover:text-primary transition-colors">
            {category.name}
          </CardTitle>
          <CardDescription className="line-clamp-2">
            {category.description}
          </CardDescription>
        </CardHeader>
        <CardContent className="text-center">
          <Badge variant="secondary" className="text-sm">
            {category.count} шпаргалок
          </Badge>
        </CardContent>
      </Card>
    </Link>
  );
};

export default CategoryCard;'''

with open(project_root / 'src/components/cheatsheet/category-card.tsx', 'w', encoding='utf-8') as f:
    f.write(category_card_component)

# CheatSheet компонентів індекс
cheatsheet_index = '''export { default as CheatSheetCard } from './cheatsheet-card';
export { default as CheatSheetContent } from './cheatsheet-content';
export { default as CheatSheetGrid } from './cheatsheet-grid';
export { default as CodeBlock } from './code-block';
export { default as SimilarCheatSheets } from './similar-cheatsheets';
export { default as CategoryCard } from './category-card';'''

with open(project_root / 'src/components/cheatsheet/index.ts', 'w', encoding='utf-8') as f:
    f.write(cheatsheet_index)

# 5. Створюємо головний індекс компонентів
main_components_index = '''// UI Components
export * from './ui';

// Layout Components
export * from './layout';

// Search Components
export * from './search';

// CheatSheet Components
export * from './cheatsheet';'''

with open(project_root / 'src/components/index.ts', 'w', encoding='utf-8') as f:
    f.write(main_components_index)

# 6. Створюємо сторінки Next.js App Router

# Root layout
root_layout = '''import type { Metadata } from 'next';
import { Inter } from 'next/font/google';
import { Layout } from '@/components/layout';
import '@/styles/globals.css';

const inter = Inter({ subsets: ['latin'] });

export const metadata: Metadata = {
  title: 'CheatSheet Hub - Шпаргалки для розробників',
  description: 'Колекція корисних шпаргалок з Docker, Git, Linux, хмарних технологій та програмування',
  keywords: ['cheatsheet', 'шпаргалки', 'docker', 'git', 'linux', 'cloud', 'programming'],
  authors: [{ name: 'CheatSheet Hub Team' }],
  openGraph: {
    title: 'CheatSheet Hub - Ваш надійний помічник у розробці',
    description: 'Знайдіть потрібні команди та приклади коду за лічені секунди',
    type: 'website',
    locale: 'uk_UA',
  },
  twitter: {
    card: 'summary_large_image',
    title: 'CheatSheet Hub',
    description: 'Шпаргалки для розробників та системних адміністраторів',
  },
};

export default function RootLayout({
  children,
}: {
  children: React.ReactNode;
}) {
  return (
    <html lang="uk" suppressHydrationWarning>
      <body className={inter.className}>
        <Layout>
          {children}
        </Layout>
      </body>
    </html>
  );
}'''

with open(project_root / 'src/app/layout.tsx', 'w', encoding='utf-8') as f:
    f.write(root_layout)

# Home page
home_page = '''import React from 'react';
import Link from 'next/link';
import { Button } from '@/components/ui/button';
import { Card, CardContent, CardDescription, CardHeader, CardTitle } from '@/components/ui/card';
import { SearchBar } from '@/components/search';
import { CategoryCard, CheatSheetGrid } from '@/components/cheatsheet';
import { Search, Zap, Heart, Download } from 'lucide-react';
import { cheatSheetService } from '@/lib/data-utils';

export default function HomePage() {
  // Get data for the home page
  const categories = cheatSheetService.categories;
  const recentCheatSheets = cheatSheetService.getAllCheatSheets().slice(0, 6);
  const stats = cheatSheetService.getStatistics();

  return (
    <div className="space-y-12">
      {/* Hero Section */}
      <section className="text-center py-12 bg-gradient-to-b from-background to-muted/20">
        <div className="container max-w-4xl">
          <h1 className="text-4xl md:text-6xl font
````
